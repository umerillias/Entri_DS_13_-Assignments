```python
# Visualization with Seaborn and Matplotlib

import matplotlib.pyplot as plt
import seaborn as sns

# Set style for all plots
sns.set_theme(style="whitegrid")

# Load the Titanic dataset
titanic = sns.load_dataset("titanic")
```

```python
# 1. Countplot: Passenger Count by Embarked Port

plt.figure(figsize=(8, 5))
sns.countplot(data=titanic, x="embarked", hue="sex", palette="muted")
plt.title("Passenger Count by Embarked Port and Sex")
plt.xlabel("Port of Embarkation (S = Southampton, C = Cherbourg, Q = Queenstown)")
plt.ylabel("Passenger Count")
plt.legend(title="Sex")
plt.show()
```

```python
# Key Observations:
# . Southampton (S): Highest passenger volume
# . Southampton (S): Heavily dominated by males
# . Queenstown (Q): Lowest overall passenger count
# . Queenstown (Q): Balanced male-to-female ratio
# . Cherbourg (C): Second most popular port
```

```python
# 2. Swarmplot: Age Distribution across Classes

plt.figure(figsize=(10, 6))
sns.swarmplot(data=titanic, x="pclass", y="age", hue="sex", palette="deep", size=4)
plt.title("Age Distribution Across Passenger Classes by Sex")
plt.xlabel("Passenger Class (Pclass)")
plt.ylabel("Age")
plt.legend(title="Sex")
plt.show()
```

```python
# Key Observations:
# . Children: Evenly spread across lower classes.
# . First Class (1): Widest age spread
# . First Class (1): Older average passenger population
# . Third Class (3): Dense clustering of youth
# . Third Class (3): Highly populated by young males
```

```python
#3. Pie Chart: Gender Distribution 

plt.figure(figsize=(6, 6))
gender_counts = titanic["sex"].value_counts()
plt.pie(
    gender_counts,
    labels=gender_counts.index,
    autopct="%1.1f%%",
    explode=[0.1, 0],
    colors=["#4c72b0", "#c44e52"],
    startangle=90,
)
plt.title("Gender Distribution of Titanic Passengers")
plt.show()
```

```python
# Key Observations:
# . Male Dominance: Roughly 64.8% of passengers.
# . Female Minority: Roughly 35.2% of passengers.
# . Gender Gap: Males outnumbered females nearly 2:1.
```

```python
# 4. Heatmap: Correlation Matrix

plt.figure(figsize=(8, 6))
columns_to_correlate = ["age", "fare", "pclass", "sibsp", "parch", "survived"]
correlation_matrix = titanic[columns_to_correlate].corr()

sns.heatmap(correlation_matrix, annot=True, cmap="coolwarm", fmt=".2f", vmin=-1, vmax=1)
plt.title("Correlation Matrix of Titanic Numerical Variables")
plt.show()
```

```python
# Key Observations:
# . Pclass & Fare: Strongest negative correlation (-0.55).
# . Pclass & Survived: Negative correlation (-0.34).
# . SibSp & Parch: Positive correlation (+0.41).
# . Age & Pclass: Negative correlation (-0.37).
# . Fare & Survived: Positive correlation (+0.26).
```

```python
# 5. Violin Plot: Fare by Class and Gender

plt.figure(figsize=(10, 6))
sns.violinplot(
    data=titanic, x="pclass", y="fare", hue="sex", palette="pastel", split=True
)
plt.title("Fare Distribution by Passenger Class and Sex")
plt.xlabel("Passenger Class (Pclass)")
plt.ylabel("Fare")
plt.show()
```

```python
# Key Observations:
# . First Class (1): Extremely long upward tails.
# . First Class (1): Highest extreme fare outliers.
# . Third Class (3): Highly compressed flat distribution.
# . Third Class (3): Uniformly cheap ticket pricing.
# . Gender Differences: Females paid higher peak fares.
```

```python
# 6. Scatter Plot: Age vs Fare

plt.figure(figsize=(10, 6))
sns.scatterplot(data=titanic, x="age", y="fare", hue="survived", palette="Set1", alpha=0.8)
plt.title("Passenger Age vs Fare Colored by Survival Status")
plt.xlabel("Age")
plt.ylabel("Fare")
plt.legend(title="Survived (0 = No, 1 = Yes)")
plt.show()
```

```python
# Key Observations:
# . High Fares: Near-perfect survival rate cluster.
# . Low Fares: Heavy concentration of non-survival dots.
# . Age Extremes: Infants show high survival rates.
# . Outliers: Rare passengers paid over $500.
```
